## Function Definition

Defining a function called split_dat_file which grabs the input file and separates out each of the columns, and saves a file for each column, containint the frequency and magnitude.
1.   Loads the data
2.   Since the .dat file that you feed this function is structured like this: **Frequency, Mag1, Mag2, ....., MagN** the first thing that the function does is separate out the **Frequency** column (index 0 = column 1).
3. Next, the function grabs the base name of the file, removing the file extension, and saves that to use later when it is saving each individual **Magnitude** column.
4. The function that creates a loop using the shape of the data that was loaded. example.shape[0] = number of rows, example.shape[1] = number of columns. Loop starts at index 1 because we already know index 0 is the frequency column. For every iteration, the loop creates a output file name, and saves that using np.savetxt formating with the frequency and magnitude, setting the number format, the delimiter, and comments.
5. Lets the user know every time it saves and output file

```python
def split_dat_file(input_file):
    """Reads a .dat file with multiple magnitude columns and splits it into separate files.
    Each output file contains frequency in the first column and one magnitude column.
    """

    # Load the .dat file, assuming tab or space delimiters
    data = np.loadtxt(input_file)

    # Extract frequency column (first column)
    frequency = data[:, 0] ##python index starts at 0

    # Get the base filename (without extension)
    base_filename = input_file.rsplit(".", 1)[0]

    # Loop through magnitude columns (starting from the second column)
    for i in range(1, data.shape[1]):
        magnitude = data[:, i]  # Extract magnitude column

        # Create output filename (e.g., inputfile_column1.dat, inputfile_column2.dat, etc.)
        output_file = f"{base_filename}_col{i}.dat"

        # Save the new .dat file
        np.savetxt(output_file, np.column_stack((frequency, magnitude)), fmt="%.6f", delimiter="\t", comments="")

        print(f"Saved: {output_file}")
```
